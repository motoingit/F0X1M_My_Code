```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Student study data
data = {
    "Hours": [
        1.5, 2.0, 2.5, 3.0, 3.5,
        4.0, 4.5, 5.0, 5.5, 6.0,
        6.5, 7.0, 7.5, 8.0, 8.5,
        9.0, 3.2, 4.8, 6.2, 7.8
    ],

    "Marks": [
        48, 55, 58, 63, 61,
        70, 68, 74, 72, 79,
        81, 84, 82, 88, 86,
        94, 65, 76, 78, 85
    ]
}

# Create DataFrame
df = pd.DataFrame(data)
```

```python
# Create regression plot
sns.regplot(
    x="Hours",
    y="Marks",
    data=df,
    scatter_kws={"s": 60},
    line_kws={"linewidth": 2}
)

# Add labels and title
plt.xlabel("Hours of Study")
plt.ylabel("Marks")
plt.title("Study Hours vs Marks")

# Display graph
plt.show()
```
