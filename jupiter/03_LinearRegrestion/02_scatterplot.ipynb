```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
# Student study data
data = {
    "Hours": [1.5, 2.0, 2.5, 3.0, 3.5,
              4.0, 4.5, 5.0, 5.5, 6.0,
              6.5, 7.0, 7.5, 8.0, 8.5,
              9.0, 2.8, 4.7, 6.8, 7.2],

    "Marks": [52, 55, 58, 61, 64,
              67, 70, 72, 75, 78,
              80, 82, 84, 87, 89,
              92, 60, 71, 81, 83]
}
```

```python
# Convert dictionary into DataFrame
df = pd.DataFrame(data)
```

```python
# Create scatter plot
ax = sns.scatterplot(x="Hours", y="Marks", data=df)

# Customize the graph
ax.set(
    xlabel="Hours of Study",
    ylabel="Marks",
    title="Study Hours vs Marks"
)

plt.show()
```
