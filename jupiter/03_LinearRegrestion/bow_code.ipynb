```python
import spacy
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
# Documents
corpus = [
  "the cat sat",
  "the cat sat on the mat",
  "the cat and another cat sat on the mat with a hat"
]
```

```python
# Lemmatization + stop-word removal
cleaned_corpus = []

for text in corpus:
    doc = nlp(text)

    words = [
        token.lemma_
        for token in doc
        if not token.is_stop and not token.is_punct
    ]

    cleaned_corpus.append(" ".join(words))
```

```python
# Show preprocessed text
print("Cleaned Corpus:")
for i, text in enumerate(cleaned_corpus, 1):
    print(f"Doc {i}: {text}")
```

```python
# Convert text into Bag of Words
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(cleaned_corpus)
```

```python
# Get vocabulary
features = vectorizer.get_feature_names_out()
```

```python
# Display BoW matrix
bow_df = pd.DataFrame(
    X.toarray(),
    columns=features,
    index=["Doc 1", "Doc 2", "Doc 3"]
)
```

```python
print("\nBag of Words Matrix:")
print(bow_df)
```
