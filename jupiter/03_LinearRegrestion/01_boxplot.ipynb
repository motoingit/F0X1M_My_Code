```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
data = {
    "Subject": [
        "Python", "Python", "Python", "Python", "Python",
        "AI", "AI", "AI", "AI", "AI",
        "DBMS", "DBMS", "DBMS", "DBMS", "DBMS"
    ],

    "Marks": [
        75, 82, 90, 78, 85,
        80, 85, 92, 88, 76,
        80, 85, 92, 89, 78
    ]
}
```

```python
df = pd.DataFrame(data)
```

```python
sns.boxplot(x="Subject", y="Marks", data=df)
plt.title("Mark Distribution")
plt.show()

```
