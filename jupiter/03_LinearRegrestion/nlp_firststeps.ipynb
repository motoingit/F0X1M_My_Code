```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
doc = nlp("I saw a movie. The saw is sharp")
print(f"Token\t\tLemma\t\tStopword".format('Token','Lemma','Stopword'))
print('-'*40)
for token in doc:
    print(f"{str(token)}\t\t{token.lemma_}\t\t{token.is_stop}")
```

```python
doc = nlp("I was not in a mood to go to school today. Today is 14 August of the year 2025.")
for token in doc:
    print(token.pos_, token.tag_, spacy.explain(token.tag_))
```

```python
doc = nlp("Use Google to google about it")
for token in doc:
    print(token.pos_, token.tag_, spacy.explain(token.tag_))
```

```python
doc = nlp("It was done at Bengaluru today. It was done at Bangalore today.")
for ent in doc.ents:
    print(ent.text,ent.label_)
```

```python
from spacy import displacy
displacy.render(doc, style="ent", jupyter="true")
```
