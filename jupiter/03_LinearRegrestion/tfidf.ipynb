```python
import spacy
from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
```

```python
nlp = spacy.load("en_core_web_sm")
```

```python
corpus = [
    "The cat runs fast",
    "The dog runs",
    "A cat is fast"
]
```

```python
lemmatized_corpus = []
for doc in corpus:
    spacy_doc = nlp(doc)
    tokens = [token.lemma_ for token in spacy_doc if not token.is_stop and not token.is_punct]
    lemmatized_corpus.append(" ".join(tokens))
```

```python
vectorizer_tfidf = TfidfVectorizer()
tfidf_matrix = vectorizer_tfidf.fit_transform(lemmatized_corpus)  
feature_names_tfidf = vectorizer_tfidf.get_feature_names_out()
tfidf_df = pd.DataFrame(tfidf_matrix.toarray(), columns=feature_names_tfidf, index=["Doc 1", "Doc 2", "Doc 3"])
print("TF-IDF Matrix:")
print(tfidf_df)
```
